Dataset Link -> https://archive.ics.uci.edu/dataset/296/diabetes+130-us+hospitals+for+years+1999-2008

```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine,text
from pathlib import Path
```

```python
engine = create_engine("mysql+pymysql://root:721401@localhost:3306/hospital_readmission")
```

```python
dataset_path = Path("dataset")

for file_path in dataset_path.glob("*.csv"):
    df = pd.read_csv(file_path)
    print(df.shape)
    table_name = file_path.stem.lower() + "_raw"
    print(table_name)

    df.to_sql(
        name = table_name,
        con=engine,
        if_exists='replace',
        index=False
    )
    print(f"Data uploaded from {file_path.name} to {table_name}")
print("ALL files are Imported")
```
